# **Сам скрипт**

```python
import telebot
from rembg import remove
from telebot.types import ReactionTypeEmoji
import random
import glob
import time
import os

bot = telebot.TeleBot('TOKEN')
@bot.message_handler(commands=['start'])
def start(message):
    bot.reply_to(message, "Привет! Я Телеграм-бот, который удаляет фон с картинок! Просто отправь мне фото, а я удалю с него весь фон.")
@bot.message_handler(content_types=['photo'])
def photo_dwnld_and_send(message):
  file_info = bot.get_file(message.photo[-1].file_id)
  file_name = file_info.file_path.split('/')[-1]
  downloaded_file = bot.download_file(file_info.file_path)
  with open(file_name, 'wb') as new_file:
      new_file.write(downloaded_file)
  files = glob.glob('/content/*.png') + glob.glob('/content/*.jpg')
  if files:
    latest_image = max(files, key=os.path.getmtime)
    with open(latest_image, 'rb') as i:
      with open('output.png', 'wb') as o:
          input = i.read()
          output = remove(input)
          o.write(output)
          with open('output.png', 'rb') as outimg:
            bot.send_photo(message.chat.id, outimg)

bot.polling()
```
